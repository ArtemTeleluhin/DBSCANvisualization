```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from json import dump
sns.set(style='whitegrid')

np.random.seed(0)
```

```python
datasets = dict()
```

```python
def scaler(data, min_x, max_x, min_y, max_y):
  x = data[:, 0]
  y = data[:, 1]
  x = (x - min_x) / (max_x - min_x)
  x = x * 800
  x = np.round(x).astype(np.int64)
  x += 100
  y = (y - min_y) / (max_y - min_y)
  y = y * 800
  y = np.round(y).astype(np.int64)
  y = 820 - y
  return [[int(i), int(j)] for i, j in zip(x, y)]
```

### Набор 1

```python
from sklearn.datasets import make_blobs

X, Y = make_blobs(
    n_samples=200,
    centers=[(16, 4), (6, 14), (11, 10), (3, 5)],
    random_state=42
)

plt.figure(figsize=(6, 6))
plt.scatter(X[:, 0], X[:, 1], alpha=0.8)
plt.xlim(0, 20)
plt.ylim(0, 20)
plt.title('Результаты опроса студентов')
plt.xlabel('Нагрузка по гуманитарным дисциплинам')
plt.ylabel('Нагрузка по техническим дисциплинам')
plt.show()
```

```python
X = scaler(X, 0, 20, 0, 20)
plt.figure(figsize=(6, 6))
plt.scatter([col[0] for col in X], [col[1] for col in X], alpha=0.8)
plt.show()
```

```python
datasets['set_1'] = X
```

### Набор 5

```python
from sklearn.datasets import make_regression

theta = np.radians(30)
c, s = np.cos(theta), np.sin(theta)
R = np.array(((c, -s), (s, c)))

X = []
for x0, y0, sz in [(0, 0, 20), (2, 0, 17), (1, 1.5, 15)]:
  x = np.random.uniform(0, 3, size=sz)
  y = np.random.uniform(0, 0.3, size=sz)
  cluster = np.column_stack((x, y)).T
  cluster = R @ cluster
  cluster = cluster.T
  cluster[:, 0] += x0
  cluster[:, 1] += y0
  X.append(cluster)
X = np.vstack(X)

plt.figure(figsize=(6, 6))
plt.scatter(X[:, 0], X[:, 1], alpha=0.8)
plt.show()
```

```python
X = scaler(X, -1, 5, -1, 5)
plt.figure(figsize=(6, 6))
plt.scatter([col[0] for col in X], [col[1] for col in X], alpha=0.8)
plt.xlim(0, 1000)
plt.ylim(0, 1000)
plt.show()
```

```python
datasets['set_5'] = X
```

```python
N = 10
x = 2 * np.arange(0, N) / N
x = np.tile(x, 3)
y = np.array([0.0, 1.0, 2.0])
y = np.repeat(y, N)

xr = x / np.sqrt(2) - y / np.sqrt(2)
yr = x / np.sqrt(2) + y / np.sqrt(2)

X = np.column_stack((np.concatenate((xr, x + 4)), np.concatenate((yr, y))))

plt.figure(figsize=(6, 6))
plt.scatter(X[:, 0], X[:, 1], alpha=0.8)
plt.show()
```

```python
X = scaler(X, -2, 6, -5, 3)
plt.figure(figsize=(6, 6))
plt.scatter([col[0] for col in X], [col[1] for col in X], alpha=0.8)
plt.xlim(0, 1000)
plt.ylim(0, 1000)
plt.show()
```

```python
datasets['set_9'] = X
```

```python
for key, value in datasets.items():
  print(key, len(value))
```
